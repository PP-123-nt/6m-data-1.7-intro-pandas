## Exercise 1: The Inventory System

```python
import pandas as pd
# {} dictionary: "key": [values...]. Each key become column name, list its values.
fruit_data = {
    "fruit": ["Apple", "Banana", "Cherry", "Date"],
    "quantity": [100, 40, 200, 20]
}

# 1. Create DataFrame
inventory = pd.DataFrame(fruit_data)

# 2. Add Price — new list assigned, one value per row, in same order.
inventory['price'] = [3.0, 4.0, 2.0, 5.0]

# 3. Calculate Total Value — multiply two columns row-by-row, store as new column.
inventory['total_value'] = inventory['quantity'] * inventory['price']

# 4. Flag Low Stock — comparison < make True/False per row.
inventory['low_stock'] = inventory['quantity'] < 50

inventory
```

## Exercise 2: Data Detective

```python
import numpy as np
# np.arange(16) → numbers 0-15. .reshape((4,4)) fold into 4 rows x 4 cols grid.
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=["Ohio", "Colorado", "Utah", "New York"],
                    columns=["one", "two", "three", "four"])

# 1. Utah and New York — .loc[[list]] select multiple rows by label at once.
print(data.loc[["Utah", "New York"]])

# 2. 'two' for 'Ohio' — .loc[row, col] pick single cell by row label + column label.
print(data.loc["Ohio", "two"])

# 3. Filter 'one' at least 8 — condition inside [] keep only rows where it's True. >= mean "at least".
print(data[data["one"] >= 8])

# 4. Update value — .loc[row, col] = new_value overwrite that one cell.
data.loc["New York", "four"] = 100
print(data)
```

## Exercise 3: Leaderboard Logic

```python
scores = pd.DataFrame({
    "Student": ["Alice", "Bob", "Charlie", "David"],
    "Math": [85, 45, 92, 60],
    "Science": [70, 55, 88, 72]
})

# 2. Sort by Math desc — ascending=False flip order, biggest first.
scores_sorted = scores.sort_values(by="Math", ascending=False)

# 3. Rank — .rank() give each value its rank position (1 = best, since ascending=False here).
scores['Math_Rank'] = scores['Math'].rank(ascending=False)

# 4 & 5. Apply Pass/Fail
# def = define a function: reusable named block of code, run later via check_status(x).
# score is placeholder name for whatever value get passed in when called.
def check_status(score):
    if score >= 60:      # if/else: run one branch or other, based on condition
        return "Pass"
    else:
        return "Fail"

# .apply(check_status) run function on every value in 'Math' column, one by one.
scores['Status'] = scores['Math'].apply(check_status)

scores
```
