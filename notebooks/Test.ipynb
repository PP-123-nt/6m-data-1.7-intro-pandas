```python
import pandas as pd
import numpy as np
# [DEMO] Detecting missing values
# np.nan = "Not a Number", pandas/numpy's way of marking missing/empty data.
data_with_nan = pd.Series([1, np.nan, 3.5, np.nan, 7])

print("Original data:")
display(data_with_nan)

print("\nWhich values are missing?")
display(data_with_nan.isna())  # isna() → True where value is NaN

print("\nWhich values are NOT missing?")
display(data_with_nan.notna())  # notna() → opposite of isna()

data_filled_zero = data_with_nan.fillna(0) # replace with 0

print("Filled with 0:")
display(data_filled_zero)
```

```python
marks = pd.DataFrame({
    "Student": ["Alice", "Bob", "Charlie", "David"],
    "Math": [85, 45, 92, 60],
    "Science": [70, 85, 88, 72]
})

def check_status(marks):
    if marks >= 60:      # if/else: run one branch or other, based on condition
        return "Pass"
    else:
        return "Fail"

# .apply(check_status) run function on every value in 'Math' column, one by one.
marks['Status'] = marks['Math'].apply(check_status)

marks
```

```python
# [DEMO] Detecting missing values
# np.nan = "Not a Number", pandas/numpy's way of marking missing/empty data.
data_with_nan = pd.Series([1, np.nan, 3.5, np.nan, 7])

print("Original data:")
display(data_with_nan)

print("\nWhich values are missing?")
display(data_with_nan.isna()) # isna() → True where value is NaN

print("\nHow many missing values?")
display(data_with_nan.isna().sum())  # isna() → True where value is NaN
```
