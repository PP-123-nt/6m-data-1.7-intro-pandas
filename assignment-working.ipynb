## Part 2: Practical Challenge

Scenario: "The Bookshop Analyst"
You're a data analyst for an online bookshop. The operations team has exported their inventory to a CSV and they need your help making sense of it.

**Challenge 1: Build and Inspect the Dataset**

Create the following DataFrame and perform a health check:

Tasks:

1. Use .info() — how many non-null values does each column have?

2. Use .describe() — what's the average price and maximum units sold?

3. Add a revenue column: price × units_sold.

4. What is the total revenue across all books?

```python
import pandas as pd

books = pd.DataFrame({
    'title': ['The Pragmatic Programmer', 'Clean Code', 'Python Crash Course',
              'Data Science from Scratch', 'Designing Data-Intensive Apps'],
    'author': ['Hunt & Thomas', 'Robert Martin', 'Eric Matthes',
               'Joel Grus', 'Martin Kleppmann'],
    'genre': ['Tech', 'Tech', 'Tech', 'Data', 'Data'],
    'price': [45.99, 38.50, 29.99, 42.00, 55.00],
    'units_sold': [121, 95, 200, 80, 60],
    'in_stock': [True, True, False, True, False]
})
display(books)
```

```python
print(books.info())  # non-null value -> with acual data (null value -> missing data)
```

```python
# summary statistics for numeric columns
display(books.describe())  

print(f"Average Price: £{books['price'].mean():,.2f}")
print(f"Max Unit Sold: {books['units_sold'].max()}")
```

```python
# add revenue column to the DataFrame and calculate total revenue
books ['revenue'] = (books['price'] * books['units_sold']).astype(float)
display(books)

print(f"Total revenue: £{books['revenue'].sum():,.2f}") # ,.2f -> commas as thousands separators & 2 decimal
```

**Challenge 2: Indexing & Filtering**

Tasks:

1. Use .loc to select only title and revenue for Data genre books.

2. Use .iloc to retrieve the last 3 rows and first 3 columns.

3. Filter books that are in stock AND have a price above £40.

4. Find the book with the highest revenue using .loc with .idxmax().

```python
# select title & revenue for 'Data' genre book
display(books[['title', 'revenue']].loc[books['genre'] == 'Data']) 

display(books.loc[books['genre'] == 'Data', ['title', 'revenue']])  # alternative way 
```

```python
# retrieve last 3 rows & first 3 columns
display(books.iloc[-3:, 0:3])  
```

```python
# select books in stock with price > £40.00
display(books.loc[(books['in_stock'] == True) & (books['price'] > 40.00)]) 
```

```python
# select book with the highest revenue
display(books.loc[books['revenue'].idxmax()])  # loc with idxmax() to get the row with the max revenue
print("\n")

top_book = books.loc[books['revenue'].idxmax()]
print(f"Top earner: {top_book['title']} — £{top_book['revenue']:,.2f}")

display(books.iloc[books['revenue'].argmax()]) # iloc with argmax() to get the row with the max revenue

display(books.sort_values('revenue').tail(1))  # sort by revenue and display the last (highest) row

display(books.nlargest(1, 'revenue'))   # display the row with the largest revenue value
```

**Challenge 3: Transform, Sort & Rank**

Tasks:

1. Use .apply() to add a price_tier column: 'Budget' (< £35), 'Standard' (£35–£50), 'Premium' (> £50).

2. Sort the DataFrame by revenue descending.

3. Add a sales_rank column using .rank(ascending=False).

4. Use .map() to encode genre as a number: {'Tech': 0, 'Data': 1}.

```python
# add price tier
def price_tier(price):
    return 'Budget' if price < 35 else 'Premium' if price > 50 else 'Standard'

display(books.assign(price_tier=books['price'].apply(price_tier)))  # add price tier column using apply() 

books['price_tier'] = books['price'].apply(price_tier)
display(books)
```

```python
# sort df by revenue in descending order
display(books.sort_values('revenue', ascending=False))
```

```python
# add sales rank column
books['sales_rank'] = books['units_sold'].rank(ascending=False).astype(int)
display(books)
```

```python
# encode genre column to numeric values
books['genre_encoded'] = books['genre'].map({'Tech': 0, 'Data': 1})
display(books)
```
